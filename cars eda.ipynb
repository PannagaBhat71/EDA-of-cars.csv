```python
chevy = df[df["Company"] == "Chevrolet"]
chevy.groupby(["Company"])["Model"].value_counts()
```

```python
def car_model(car_name):
    name = df[df["Company"] == f"{car_name}"]
    return name.groupby(["Company"])["Model"].value_counts()


car_model("Audi")
```
